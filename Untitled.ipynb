```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ==========================
# LOAD DATA
# ==========================

df = pd.read_csv("/content/web_traffic.csv")

print("Dataset Shape:")
print(df.shape)

print("\nFirst 5 Records")
print(df.head())
```

```python
# ==========================
# DATA CLEANING
# ==========================

df.drop_duplicates(inplace=True)

for col in df.select_dtypes(include=np.number):
    df[col].fillna(df[col].median(), inplace=True)

for col in df.select_dtypes(include='object'):
    df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing Values")
print(df.isnull().sum())
```

```python
# ==========================
# PAGE VIEWS ANALYSIS
# ==========================

plt.figure(figsize=(8,5))
sns.histplot(df['PagesViewed'], bins=20)
plt.title("Page Views Distribution")
plt.show()
```

```python
# ==========================
# SESSION DURATION
# ==========================

plt.figure(figsize=(8,5))
sns.histplot(df['SessionDuration'], bins=20)
plt.title("Session Duration Distribution")
plt.show()
```

```python
# ==========================
# TRAFFIC SOURCE ANALYSIS
# ==========================

plt.figure(figsize=(8,5))
sns.countplot(x='TrafficSource', data=df)
plt.title("Traffic Sources")
plt.xticks(rotation=30)
plt.show()
```

```python
# ==========================
# DEVICE ANALYSIS
# ==========================

plt.figure(figsize=(8,5))
sns.countplot(x='DeviceType', data=df)
plt.title("Device Usage")
plt.show()
```

```python
# ==========================
# BOUNCE RATE
# ==========================

bounce_rate = (
    df['Bounce'].value_counts()['Yes']
    / len(df)
) * 100

print(f"\nBounce Rate: {bounce_rate:.2f}%")
```

```python
# ==========================
# LANDING PAGE ANALYSIS
# ==========================

print("\nTop Landing Pages")
print(
    df['LandingPage']
    .value_counts()
    .head(10)
)
```

```python
# ==========================
# EXIT PAGE ANALYSIS
# ==========================

print("\nTop Exit Pages")
print(
    df['ExitPage']
    .value_counts()
    .head(10)
)
```

```python
# ==========================
# USER JOURNEY
# ==========================

journey = pd.crosstab(
    df['LandingPage'],
    df['ExitPage']
)

print("\nUser Journey Matrix")
print(journey)
```

```python
# ==========================
# CONVERSION RATE
# ==========================

conversion_rate = (
    df['Conversion'].value_counts()['Yes']
    / len(df)
) * 100

print(f"\nConversion Rate: {conversion_rate:.2f}%")
```

```python
# ==========================
# SESSION DURATION VS CONVERSION
# ==========================

plt.figure(figsize=(8,5))
sns.boxplot(
    x='Conversion',
    y='SessionDuration',
    data=df
)
plt.title("Session Duration vs Conversion")
plt.show()
```

```python
# ==========================
# PAGE VIEWS VS CONVERSION
# ==========================

plt.figure(figsize=(8,5))
sns.boxplot(
    x='Conversion',
    y='PagesViewed',
    data=df
)
plt.title("Pages Viewed vs Conversion")
plt.show()
```

```python
# ==========================
# SAVE CLEAN DATA
# ==========================

df.to_csv(
    "clean_web_traffic.csv",
    index=False
)

print("\nClean dataset saved.")
```
